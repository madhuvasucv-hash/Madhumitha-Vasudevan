```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# Create dataset
data = {
    'RAM': [2, 3, 4, 4, 6, 8, 8, 12, 6, 2, 12, 8],
    'Storage': [16, 32, 64, 64, 128, 128, 256, 256, 128, 16, 512, 256],
    'Battery': [2000, 2500, 3000, 3500, 4000, 4500,
                5000, 5000, 4500, 2200, 6000, 5000],
    'Camera': [5, 8, 12, 13, 16, 20, 32, 48, 24, 5, 64, 48],
    'Price_Range': [0, 0, 1, 1, 2, 2, 3, 3, 2, 0, 3, 3]
}

df = pd.DataFrame(data)

# Input and output
X = df[['RAM', 'Storage', 'Battery', 'Camera']]
y = df['Price_Range']

# Split dataset (without stratify)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

print("Actual Values:", y_test.values)
print("Predicted Values:", y_pred)

print("Accuracy:", accuracy_score(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Predict a new mobile
new_mobile = pd.DataFrame(
    [[8, 256, 5000, 48]],
    columns=['RAM', 'Storage', 'Battery', 'Camera']
)

result = model.predict(new_mobile)

categories = {
    0: "Low Cost",
    1: "Medium Cost",
    2: "High Cost",
    3: "Very High Cost"
}

print("Mobile Price Prediction:", categories[result[0]])
```
