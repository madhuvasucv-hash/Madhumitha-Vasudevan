```python
# Candidate-Elimination Algorithm in Python

import numpy as np

# Training dataset: [Sky, AirTemp, Humidity, Wind, Water, Forecast, EnjoySport]
training_data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

num_attributes = len(training_data[0]) - 1

# Initialize Specific Boundary (S) with most specific hypothesis ('0')
S = ['0'] * num_attributes

# Initialize General Boundary (G) with most general hypothesis ('?')
G = [['?'] * num_attributes]

print("Initial Boundaries:")
print("S0:", S)
print("G0:", G)
print("-" * 50)

# Helper function to check if a hypothesis more_general_than or matches an example
def matches(hypothesis, example):
    for i in range(len(hypothesis)):
        if hypothesis[i] != '?' and hypothesis[i] != example[i]:
            return False
    return True

# Helper function for generalization/specialization checks
def fulfills(hypothesis, example):
    for i in range(len(hypothesis)):
        if hypothesis[i] != '?' and hypothesis[i] != '0' and hypothesis[i] != example[i]:
            return False
    return True

# Candidate-Elimination Algorithm Steps
for index, example in enumerate(training_data):
    x = example[:-1]
    label = example[-1]

    if label == 'Yes':  # Positive Example
        # 1. General Boundary: Remove any hypothesis in G that is inconsistent with x
        G = [g for g in G if fulfills(g, x)]

        # 2. Specific Boundary: Update S
        for i in range(num_attributes):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        # Remove any hypotheses from S that are more general than any hypothesis in G (optional cleanup)
        # and ensure S is consistent with G
        for g in G:
            for i in range(num_attributes):
                if S[i] != '?' and S[i] != g[i] and g[i] != '?':
                    # Handle specialization if needed
                    pass

    else:  # Negative Example
        # 1. Specific Boundary: Remove any hypothesis in S inconsistent with x
        # (For Candidate-Elimination, negative examples prune S, though standard S stays consistent)

        # 2. General Boundary: Update G
        new_G = []
        for g in G:
            if fulfills(g, x):
                # Specialize g to exclude x while remaining consistent with S
                for i in range(num_attributes):
                    if g[i] == '?':
                        for val in set(row[i] for row in training_data):
                            if val != x[i]:
                                temp_g = list(g)
                                temp_g[i] = val
                                # Ensure temp_g is more general than S
                                if fulfills(temp_g, S):
                                    new_G.append(temp_g)
            else:
                new_G.append(g)
        G = new_G

    print(f"After Example {index + 1} ({label}):")
    print("S =", S)
    print("G =", G)
    print("-" * 30)

print("\nFinal Version Space:")
print("Specific Boundary (S):", S)
print("General Boundary (G):", G)
```

```python
import math
import pandas as pd
from collections import Counter
from pprint import pprint

# Dataset
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain',
                'Rain', 'Rain', 'Overcast', 'Sunny',
                'Sunny', 'Rain', 'Sunny', 'Overcast',
                'Overcast', 'Rain'],

    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild',
                    'Cool', 'Cool', 'Cool', 'Mild',
                    'Cool', 'Mild', 'Mild', 'Mild',
                    'Hot', 'Mild'],

    'Humidity': ['High', 'High', 'High', 'High',
                 'Normal', 'Normal', 'Normal', 'High',
                 'Normal', 'Normal', 'Normal', 'High',
                 'Normal', 'High'],

    'Wind': ['Weak', 'Strong', 'Weak', 'Weak',
             'Weak', 'Strong', 'Strong', 'Weak',
             'Weak', 'Weak', 'Strong', 'Strong',
             'Weak', 'Strong'],

    'Play': ['No', 'No', 'Yes', 'Yes',
             'Yes', 'No', 'Yes', 'No',
             'Yes', 'Yes', 'Yes', 'Yes',
             'Yes', 'No']
}

df = pd.DataFrame(data)

# Calculate entropy
def entropy(target):
    counts = Counter(target)
    total = len(target)
    result = 0

    for count in counts.values():
        p = count / total
        result -= p * math.log2(p)

    return result

# Calculate information gain
def information_gain(df, attribute, target):
    total_entropy = entropy(df[target])
    weighted_entropy = 0

    for value in df[attribute].unique():
        subset = df[df[attribute] == value]
        weighted_entropy += (
            len(subset) / len(df)
        ) * entropy(subset[target])

    return total_entropy - weighted_entropy

# Build ID3 decision tree
def id3(df, attributes, target):
    if len(df[target].unique()) == 1:
        return df[target].iloc[0]

    if not attributes:
        return df[target].mode()[0]

    best = max(
        attributes,
        key=lambda a: information_gain(df, a, target)
    )

    tree = {best: {}}

    for value in df[best].unique():
        subset = df[df[best] == value]
        remaining = [a for a in attributes if a != best]

        tree[best][value] = id3(
            subset, remaining, target
        )

    return tree

# Classification function
def predict(tree, sample):
    if not isinstance(tree, dict):
        return tree

    attribute = next(iter(tree))
    value = sample[attribute]

    if value not in tree[attribute]:
        return "Unknown"

    return predict(tree[attribute][value], sample)

# Display information gain
attributes = ['Outlook', 'Temperature',
              'Humidity', 'Wind']

for attr in attributes:
    print(attr, ":", round(
        information_gain(df, attr, 'Play'), 3
    ))

# Construct decision tree
tree = id3(df, attributes, 'Play')

print("\nDecision Tree:")
pprint(tree)

# Classify a new sample
new_sample = {
    'Outlook': 'Sunny',
    'Temperature': 'Cool',
    'Humidity': 'High',
    'Wind': 'Weak'
}

result = predict(tree, new_sample)

print("\nNew Sample:", new_sample)
print("Predicted Result:", result)
```

```python

import numpy as np

# XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of sigmoid
def sigmoid_derivative(x):
    return x * (1 - x)

# Initialize random weights
np.random.seed(42)

W1 = np.random.uniform(-1, 1, (2, 4))
W2 = np.random.uniform(-1, 1, (4, 1))

b1 = np.zeros((1, 4))
b2 = np.zeros((1, 1))

learning_rate = 0.5
epochs = 20000

# Training using backpropagation
for epoch in range(epochs):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate error
    error = Y - predicted_output

    # Backpropagation
    output_delta = error * sigmoid_derivative(
        predicted_output
    )

    hidden_error = np.dot(output_delta, W2.T)

    hidden_delta = hidden_error * sigmoid_derivative(
        hidden_output
    )

    # Update weights and biases
    W2 += learning_rate * np.dot(
        hidden_output.T, output_delta
    )

    b2 += learning_rate * np.sum(
        output_delta, axis=0, keepdims=True
    )

    W1 += learning_rate * np.dot(
        X.T, hidden_delta
    )

    b1 += learning_rate * np.sum(
        hidden_delta, axis=0, keepdims=True
    )

    # Display error
    if epoch % 5000 == 0:
        loss = np.mean(error ** 2)
        print("Epoch:", epoch, "Error:", loss)

# Test the trained network
print("\nFinal Predictions:")

for i in range(len(X)):
    h = sigmoid(np.dot(X[i:i+1], W1) + b1)
    output = sigmoid(np.dot(h, W2) + b2)

    prediction = 1 if output[0][0] >= 0.5 else 0

    print(
        X[i], "Actual:", Y[i][0],
        "Predicted:", prediction,
        "Output:", round(output[0][0], 3)
    )
```

```python

import numpy as np
from collections import Counter

# Training dataset
# Features: [Study Hours, Attendance]
X = np.array([
    [2, 60],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85]
])

# Class labels
y = np.array([
    'Fail',
    'Fail',
    'Fail',
    'Pass',
    'Pass',
    'Pass'
])

# K-NN function
def knn_predict(X, y, new_point, k):
    distances = []

    # Calculate Euclidean distance
    for i in range(len(X)):
        distance = np.sqrt(
            np.sum((X[i] - new_point) ** 2)
        )
        distances.append((distance, y[i]))

    # Sort distances
    distances.sort(key=lambda x: x[0])

    # Select K nearest neighbours
    nearest = distances[:k]

    # Find majority class
    classes = [label for distance, label in nearest]

    prediction = Counter(classes).most_common(1)[0][0]

    return prediction

# New student data
new_student = np.array([5, 78])

# Choose K value
k = 3

# Predict result
result = knn_predict(X, y, new_student, k)

print("New Student:", new_student)
print("K value:", k)
print("Predicted Result:", result)
```

```python

# Import required libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create Naive Bayes model
model = GaussianNB()

# Train the model
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Display confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy * 100, "%")

# Display classification report
print("\nClassification Report:")
print(classification_report(
    y_test, y_pred, target_names=iris.target_names
))
```
