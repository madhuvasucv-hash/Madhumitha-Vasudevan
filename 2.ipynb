```python

import csv

# Create training dataset
data = [
    ['Sky', 'Temperature', 'Humidity', 'Wind',
     'Water', 'Forecast', 'EnjoySport'],

    ['Sunny', 'Warm', 'Normal', 'Strong',
     'Warm', 'Same', 'Yes'],

    ['Sunny', 'Warm', 'High', 'Strong',
     'Warm', 'Same', 'Yes'],

    ['Rainy', 'Cold', 'High', 'Strong',
     'Warm', 'Change', 'No'],

    ['Sunny', 'Warm', 'High', 'Strong',
     'Cool', 'Change', 'Yes']
]

# Save data into CSV file
with open('training_data.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)

# Read CSV file
with open('training_data.csv', 'r') as file:
    reader = csv.reader(file)
    rows = list(reader)

attributes = rows[0][:-1]
examples = rows[1:]

# Initialize boundaries
S = ['0'] * len(attributes)
G = [['?'] * len(attributes)]

# Check if hypothesis covers example
def covers(h, x):
    return all(
        h[i] == '?' or h[i] == x[i]
        for i in range(len(h))
    )

# Candidate-Elimination algorithm
for row in examples:
    x = row[:-1]
    target = row[-1].strip().lower()

    if target == 'yes':
        # Remove inconsistent hypotheses from G
        G = [g for g in G if covers(g, x)]

        # Generalize S
        for i in range(len(S)):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        # Remove hypotheses from G that cover negative example
        new_G = []

        for g in G:
            if covers(g, x):
                for i in range(len(g)):
                    if g[i] == '?':
                        values = set(
                            r[i] for r in rows[1:]
                        )
                        for value in values:
                            if value != x[i] and S[i] in ('0', '?', value):
                                new_h = g.copy()
                                new_h[i] = value

                                if all(
                                    new_h[j] == '?' or
                                    S[j] == '?' or
                                    new_h[j] == S[j]
                                    for j in range(len(S))
                                ):
                                    new_G.append(new_h)
            else:
                new_G.append(g)

        G = new_G

    # Remove duplicate hypotheses
    G = [list(h) for h in dict.fromkeys(tuple(g) for g in G)]

# Display results
print("Attributes:", attributes)
print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)

print("\nCSV File Created Successfully!")
```
